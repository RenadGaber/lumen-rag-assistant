```python
def extract_pdf_text(file_path):
    text = ""
    with pdfplumber.open(file_path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    return text
```

```python
content = extract_pdf_text(data_path + "Lec 2 - Arrays.pdf")
```

```python
def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap
    return chunks
```

```python
all_chunks = []
all_sources = []  # ← جديد

for file in pdf_files:
    content = extract_pdf_text(data_path + file)
    file_chunks = chunk_text(content)
    all_chunks.extend(file_chunks)
    all_sources.extend([file] * len(file_chunks))  # ← نفس اسم الملف لكل جزء بتاعه

print(f"إجمالي عدد الأجزاء: {len(all_chunks)}")
```

```python
from sentence_transformers import SentenceTransformer


embedder = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

# تحويل كل الـ chunks لـ embeddings

embeddings = embedder.encode(all_chunks, show_progress_bar=True)
print(f"شكل الـ embeddings: {embeddings.shape}")
```

```python
chroma_client = chromadb.PersistentClient(path="/content/drive/MyDrive/lumen-rag/vector_store")

# احذف الـ collection القديمة الأول عشان منكررش نفس البيانات
chroma_client.delete_collection(name="lumen_docs")
collection = chroma_client.get_or_create_collection(name="lumen_docs")

collection.add(
    documents=all_chunks,
    embeddings=embeddings.tolist(),
    metadatas=[{"source": s} for s in all_sources],   # ← الإضافة المهمة
    ids=[f"chunk_{i}" for i in range(len(all_chunks))]
)

print(f"اتخزن {collection.count()} جزء مع مصادرها")
```

```python
def retrieve(query, n_results=3):
    query_embedding = embedder.encode([query]).tolist()
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=n_results
    )
    return results['documents'][0]
```

```python
def ask(query):
    context_chunks = retrieve(query)
    context = "\n\n".join(context_chunks)

    messages = [
        {
            "role": "system",
            "content": (


          "You are a helpful study assistant for Computer Science students. "
          "Answer strictly based on the given context. Use clear bullet points when comparing concepts. "
          "If the context is not enough, say the information is not in the documents. "
         "Answer in English only."
           )
        },
        {
            "role": "user",
            "content": f"السياق:\n{context}\n\nالسؤال: {query}"
        }
    ]

    response = generator(
        messages,
        max_new_tokens=350,
        do_sample=True,
        temperature=0.4,
        repetition_penalty=1.2
    )
    return response[0]['generated_text'][-1]['content']
```
